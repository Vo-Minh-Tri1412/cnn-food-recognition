# 02 - Demo Checkout

Notebook này chạy demo: ảnh khay → crop món → classify → tra giá → xuất hóa đơn JSON.

```python
from pathlib import Path
import json, subprocess, sys
from IPython.display import display, Image

ROOT = Path.cwd()
demo_images = sorted((ROOT / 'data/demo_trays').glob('*.*'))
print('Demo images:', len(demo_images))
for p in demo_images[:5]:
    print(p)
```

```python
# Pick one image. Change index if needed.
image_path = demo_images[0]
display(Image(filename=str(image_path)))
```

```python
cmd = [sys.executable, 'scripts/06_demo_checkout.py', '--image', str(image_path)]
print('running:', ' '.join(cmd))
subprocess.run(cmd, cwd=ROOT, check=True)
```

```python
bill_path = ROOT / 'outputs/bills' / f'{image_path.stem}_bill.json'
bill = json.loads(bill_path.read_text(encoding='utf-8'))
print(json.dumps(bill, indent=2, ensure_ascii=False))
```

```python
for item in bill['items']:
    crop_path = Path(item['crop_path'])
    if crop_path.exists():
        print(item['display_name'], item['confidence'], item['price_vnd'])
        display(Image(filename=str(crop_path)))
```
