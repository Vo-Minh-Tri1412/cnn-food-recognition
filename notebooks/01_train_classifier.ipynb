Notebook này train model nhận diện 11 món sau khi đã có ảnh từng món trong `data/classification/`.

````python
from pathlib import Path
import subprocess, sys

ROOT = Path.cwd()
print('Project root:', ROOT)
````

````python
import torch
print('torch:', torch.__version__)
print('cuda available:', torch.cuda.is_available())
print('device:', torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'CPU')
````

````python
def run_script(*args):
    cmd = [sys.executable, *args]
    print('running:', ' '.join(cmd))
    subprocess.run(cmd, cwd=ROOT, check=True)

run_script('scripts/01_inventory_data.py')
````

Nếu inventory báo `classification: 0 images`, cần thêm ảnh train trước khi chạy cell train. Dataset đúng format:

```text
data/classification/train/com_trang/*.jpg
data/classification/val/com_trang/*.jpg
data/classification/test/com_trang/*.jpg
```

````python
# Adjust epochs/batch-size depending on GPU memory.
run_script('scripts/05_train_classifier.py', '--epochs', '5', '--batch-size', '8')
````
